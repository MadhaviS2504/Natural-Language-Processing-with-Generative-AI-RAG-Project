```python
#uncomment the below snippet of code if the runtime is connected to GPU.
llm = Llama(
    model_path=model_path,
    n_ctx=5000,
    n_gpu_layers=38,
    n_batch=512
)
```

#### Generate, Process, Return responses from the LLM

```python
# function to generate, process, and return the response from the LLM
def generate_llama_response(user_prompt, max_tokens=1024, temperature=0.01, top_p=0.95, top_k=50, custom_system_message=None):

    # Default system message
    default_system_message = """
    [INST]<<SYS>> Respond to the user question based on the user prompt<</SYS>>[/INST]
    """
    # Use custom system message if provided, otherwise use default
    system_message = custom_system_message if custom_system_message else default_system_message

    # Combine user_prompt and system_message to create the prompt
    prompt = f"{user_prompt}\n{system_message}"

    # Generate a response from the LLaMA model
    response = llm(
        prompt=prompt,
        max_tokens=max_tokens,
        temperature=temperature,
        top_p=top_p,
        repeat_penalty=1.2,
        top_k=top_k,
        stop=['INST'],
        echo=False
    )

    # Extract and return the response text
    response_text = response["choices"][0]["text"]
    return response_text
```

```python
#lets define a custom prompt and will use this as guidance to the llm on how to answer a question.
prompt_engineered_system_message = """
[INST]<<SYS>>
Your task is to provide detailed and accurate medical information based on the user's question.
When providing treatment plans or protocols, structure your response as a numbered list of steps or recommendations.

For symptoms or causes, present them as bulleted lists.
Ensure your response is clear, concise, and directly answers all parts of the user's query.

Here is an example:
User: What are the common symptoms of a common cold?
Assistant: The common symptoms of a common cold include:
* Runny nose
* Sore throat
* Cough
* Congestion
* Slight body aches or a mild headache
* Sneezing
* Low-grade fever
* Generally feeling unwell (malaise)

Now, respond to the user's question:
<</SYS>>

[/INST]
"""
```

- Lets define a  `common_system_message_content` variable, which will be used across all different prompt engineering scenarios to ensure consistent  behavior of the model.



```python
common_system_message_content = """
Your task is to provide detailed and accurate medical information based on the user's question.
When providing treatment plans or protocols, structure your response as a numbered list of steps or recommendations.
For symptoms or causes, present them as bulleted lists.
Ensure your response is clear, concise, and directly answers all parts of the user's query.

Here is an example:
User: What are the common symptoms of a common cold?
Assistant: The common symptoms of a common cold include:
* Runny nose
* Sore throat
* Cough
* Congestion
* Slight body aches or a mild headache
* Sneezing
* Low-grade fever
* Generally feeling unwell (malaise)

Now, respond to the user's question:
"""
```

**Define the LLM parameter set**:

- Now, I will define five distinct dictionaries, each representing a set of LLM parameters (max_tokens, temperature, top_p, top_k) for tuning



```python
llm_parameter_set_1 = {
    "max_tokens": 1024,
    "temperature": 0.01,
    "top_p": 0.95,
    "top_k": 50
}

llm_parameter_set_2 = {
    "max_tokens": 768,
    "temperature": 0.1,
    "top_p": 0.9,
    "top_k": 40
}

llm_parameter_set_3 = {
    "max_tokens": 1280,
    "temperature": 0.05,
    "top_p": 0.98,
    "top_k": 60
}

llm_parameter_set_4 = {
    "max_tokens": 512,
    "temperature": 0.2,
    "top_p": 0.85,
    "top_k": 30
}

llm_parameter_set_5 = {
    "max_tokens": 1536,
    "temperature": 0.001,
    "top_p": 0.99,
    "top_k": 70
}

print("Five distinct LLM parameter sets are defined.")
```

- Now lets apply the five defined LLM parameter tuning combinations to the list of the questions and see the responses provided by the LLM:

- We will apply the tuned params as below:
    - Query 1: llm_parameter_set_1
    - Query 2: llm_parameter_set_2
    - Query 3 : llm_parameter_set_3
    - Query 4: llm_parameter_set_4
    - Query 5: llm_parameter_set_5


```python
def get_tuned_llm_responses_for_query(user_query, parameter_set, custom_system_message):

    max_tokens = parameter_set['max_tokens']
    temperature = parameter_set['temperature']
    top_p = parameter_set['top_p']
    top_k = parameter_set['top_k']

    #lets print all the defined set of parameter combinations and
    #ask the query 1 to the LLM with the settings applied for each parameter set.
    print(f"\n--- Generating Response for Parameter Set  ---")
    print(f"Parameters: max_tokens={max_tokens}, temperature={temperature}, top_p={top_p}, top_k={top_k} \n")

    response = generate_llama_response(
        user_prompt=user_query,
        custom_system_message=custom_system_message,
        max_tokens=max_tokens,
        temperature=temperature,
        top_p=top_p, top_k=top_k
    )

    return response
```

### System and User Prompt Template

Prompts guide the model to generate accurate responses. Here, we define two parts:

    1. The system message describing the assistant's role.
    2. A user message template including context and the question.

```python
qna_system_message = """
You are an assistant whose work is to review the report and provide the appropriate answers from the context.
User input will have the context required by you to answer user questions.
This context will begin with the token: ###Context.
The context contains references to specific portions of a document relevant to the user query.

User questions will begin with the token: ###Question.

Please answer only using the context provided in the input. Do not mention anything about the context in your final answer.

If the answer is not found in the context, respond "I don't know".
"""
```

```python
qna_user_message_template = """
###Context
Here are some documents that are relevant to the question mentioned below.
{context}

###Question
{question}
"""
```

### Response Function

```python
def generate_rag_response(user_input,k=3,max_tokens=1536,temperature=0,top_p=0.95,top_k=70):
    global qna_system_message,qna_user_message_template
    # Retrieve relevant document chunks
    relevant_document_chunks = retriever.get_relevant_documents(query=user_input,k=k)
    context_list = [d.page_content for d in relevant_document_chunks]

    # Combine document chunks into a single context
    context_for_query = ". ".join(context_list)

    user_message = qna_user_message_template.replace('{context}', context_for_query)
    user_message = user_message.replace('{question}', user_input)

    prompt = qna_system_message + '\n' + user_message

    # Generate the response using the same llm model that we have loded initially.
    try:
        response = llm(
                  prompt=prompt,
                  max_tokens=max_tokens,
                  temperature=temperature,
                  top_p=top_p,
                  top_k=top_k
                  )

        # Extract and print the model's response
        response = response['choices'][0]['text'].strip()
    except Exception as e:
        response = f'Sorry, I encountered the following error: \n {e}'

    return response
```

```python
user_message_template = """
###Question
{question}

###Context
{context}

###Answer
{answer}
"""
```

### Defining the Evaluation Function

```python
def generate_ground_relevance_response(user_input,k=3,max_tokens=1536,temperature=0,top_p=0.95,top_k=70):
    global qna_system_message,qna_user_message_template
    # Retrieve relevant document chunks
    relevant_document_chunks = retriever.get_relevant_documents(query=user_input,k=3)
    context_list = [d.page_content for d in relevant_document_chunks]
    context_for_query = ". ".join(context_list)

    # Combine user_prompt and system_message to create the prompt
    prompt = f"""[INST]{qna_system_message}\n
                {'user'}: {qna_user_message_template.format(context=context_for_query, question=user_input)}
                [/INST]"""

    response = llm(
            prompt=prompt,
            max_tokens=max_tokens,
            temperature=temperature,
            top_p=top_p,
            top_k=top_k,
            stop=['INST'],
            )

    answer =  response["choices"][0]["text"]
    # Combine user_prompt and system_message to create the prompt
    groundedness_prompt = f"""[INST]{groundedness_rater_system_message}\n
                {'user'}: {user_message_template.format(context=context_for_query, question=user_input, answer=answer)}
                [/INST]"""

    # Combine user_prompt and system_message to create the prompt
    relevance_prompt = f"""[INST]{relevance_rater_system_message}\n
                {'user'}: {user_message_template.format(context=context_for_query, question=user_input, answer=answer)}
                [/INST]"""

    response_1 = llm(
            prompt=groundedness_prompt,
            max_tokens=max_tokens,
            temperature=temperature,
            top_p=top_p,
            top_k=top_k,
            stop=['INST'],
            )

    response_2 = llm(
            prompt=relevance_prompt,
            max_tokens=max_tokens,
            temperature=temperature,
            top_p=top_p,
            top_k=top_k,
            stop=['INST'],
            )

    return response_1['choices'][0]['text'],response_2['choices'][0]['text']
```
